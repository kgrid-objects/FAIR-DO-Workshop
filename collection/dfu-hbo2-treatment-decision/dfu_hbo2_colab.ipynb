```python
%%bash
node - <<'NODE'
const { evaluateSerializedInput, evaluateFixture } = require('./src/decision');

const duplicateRawJson = '{"dfu_confirmed":"true","dfu_confirmed":"false","wagner_grade":3,"acute_surgical_intervention":"false","not_healed_after_30_days":"false"}';
const duplicateResult = evaluateSerializedInput(duplicateRawJson);

const absentResult = evaluateFixture({ kind: 'absent' });

console.log('evaluateSerializedInput (duplicate property) result:');
console.log(JSON.stringify(duplicateResult, null, 2));

console.log('evaluateFixture (absent) result:');
console.log(JSON.stringify(absentResult, null, 2));
NODE
```
